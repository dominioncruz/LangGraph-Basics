```python
from typing import TypedDict, List, Dict
from langchain_core.messages import BaseMessage, HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv

from typing import Annotated
from langgraph.graph.message import add_messages
```

```python
load_dotenv()
```

```python
class AgentState(TypedDict):
    messages: List[HumanMessage]
    
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite-preview")
#chose 3.1 flash lite because of it's RPMs and limit of 500 messages per day, other models have a terrible cap and i'm not ready to switch models when i hit the limit

#the downside is it's a bit slow
```

```python
def process(state: AgentState) -> dict:
    response = llm.invoke(state['messages'])
    return {"messages": [response.content]}
```

```python
graph = StateGraph(AgentState)
graph.add_node('process', process)
graph.add_edge(START, 'process')
graph.add_edge('process', END)

app = graph.compile()
```

```python
result = app.invoke({'messages': [HumanMessage(content='how are you doing?')]})
```
