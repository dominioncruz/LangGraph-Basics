```python
from typing import TypedDict, List, Dict, NotRequired, Optional, Literal, Callable

from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    operator: Literal["*", "+"]
    result: NotRequired[Optional[str]]
```

```python
import math

OPERATIONS: Dict[str, Callable[[List[int]], int]] = {
    "*": lambda lst: math.prod(lst),
    "+": lambda lst: sum(lst)
}

def process_node(state: AgentState) -> AgentState:
    """this alters result based on operator type

    Args:
        state (AgentState): _description_

    Returns:
        AgentState: _description_
    """
    result = OPERATIONS[state["operator"]](state["values"])
    state["result"] = f"Hey {state["name"].capitalize()}, after applying the operator {state["operator"]} on the list {state["values"]}, your result is {result}"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("process", process_node)
graph.set_entry_point("process")
graph.set_finish_point("process")

app = graph.compile()
```

```python
result = app.invoke({ "values": [1,2,3,4], "name": "miracle", "operator": "+"})
```
