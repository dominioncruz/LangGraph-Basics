```python
from typing import Dict, TypedDict

from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    message: str
    
    
def compliment_node(state: AgentState) -> AgentState:
    """__returns a simple compliment to the user__

    Args:
        state (AgentState): the agent state before adding message

    Returns:
        AgentState: the agent state after adding message
    """
    state["message"] = f"{state["message"].capitalize()}, you're looking baddddddd"
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("message", compliment_node)
graph.set_entry_point("message")
graph.set_finish_point("message")

app = graph.compile()
```

```python
result = app.invoke({"message": "miracle"})

result["message"]
```
