```python
from typing import TypedDict, List, Dict, NotRequired, Callable

from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    skills: List[str]
    name: str
    age: str
    result: NotRequired[str]
```

```python
def first_node(state: AgentState) -> AgentState:
    """This is the first node, for setup"""
    skills  = state["skills"]
    all_but_last = ", ".join(skills[:-1])
    last = skills[-1]
    state["result"] = f"Your skills are {all_but_last} and {last}"
    return state

def second_node(state: AgentState) -> AgentState:
    """This is the second one

    Args:
        state (AgentState): _description_
    """
    current_state = state.get("result", "")
    state["result"] = current_state + " " + state.get("name")
    return state
    
def third_node(state: AgentState) -> AgentState:
    """This is the third one

    Args:
        state (AgentState): _description_
    """
    current_state = state.get("result", "")
    state["result"] = current_state + ". " + f"Finally you are {state['age']} years old!"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("first", first_node)
graph.add_edge("first", "second")
graph.add_node("second", second_node)
graph.add_edge("second", "third")
graph.add_node("third", third_node)
graph.set_entry_point("first")
graph.set_finish_point("third")

app = graph.compile()
```

```python
result = app.invoke({ "skills": ["python", "typescript"], "name": "miracle", "age": "23"})
```
